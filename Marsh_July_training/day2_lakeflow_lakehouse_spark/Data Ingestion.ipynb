```sql
describe extended dev.naval.transactions
```

```sql
SELECT
  `location_state`, location_city,
  SUM(COALESCE(`fee_amount`, 0.0)) AS total_fee_amount
FROM `dev`.`naval`.`transactions`
GROUP BY `location_state`, location_city
ORDER BY total_fee_amount DESC;
```

```sql
create table if not exists dev.naval.transactions_metadata as 
select *, _metadata.file_path, _metadata.file_name , current_timestamp as ingestion_time from read_files("/Volumes/dev/naval/raw/transactions.csv")
```
