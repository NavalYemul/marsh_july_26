```sql
-- Clean transactions: remove metadata columns, deduplicate, add data quality checks

create or replace table IDENTIFIER(concat(:catalog, '.', 'naval_silver.transactions')) as
with cleaned_transactions as (
  select
    transaction_id,
    from_account_id,
    to_account_id,
    transaction_type,
    amount,
    transaction_date,
    transaction_timestamp,
    channel,
    merchant_name,
    merchant_category,
    description,
    reference_number,
    status,
    currency,
    exchange_rate,
    fee_amount,
    location_city,
    location_state,
    device_type,
    ip_address
  from IDENTIFIER(concat(:catalog, '.', 'naval_bronze.transactions'))
  where transaction_id is not null
    and from_account_id is not null
    and amount > 0
)
select distinct *
from cleaned_transactions
```
